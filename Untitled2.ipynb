# Exercicio 1 — Ficha de dados pessoais

* Crie um script que armazene em variaveis o seu nome, idade, altura e se voce e maior de idade
(True ou False). Use a funcao type() para imprimir na tela o tipo de cada uma dessas variaveis.
Depois, imprima uma frase unica juntando todas as informacoes, por exemplo: Meu nome e Ana,
tenho 19 anos, 1.68m de altura e sou maior de idade: True

```python
nome = input("Qual é o seu nome? ")
idade = int(input("Qual a sua idade? "))
altura = float(input("Qual é a sua altura? "))

if idade <= 18 :
  maior_de_idade = False
else :
  maior_de_idade = True

print(type(nome))
print(type(idade))
print(type(altura))
print(type(maior_de_idade))

print(f"Meu mome é {nome}, tenho {idade} anos, {altura}m de altura e sou maior de idade: {maior_de_idade}")
```

# Exercicio 9 — Divisao e resto
* Crie duas variaveis numericas inteiras representando uma quantidade total de doces e um numero
de criancas. Use o operador de divisao inteira (//) para descobrir quantos doces cada crianca
recebe, e o operador de resto (%) para descobrir quantos doces sobram. Imprima os dois
resultados.

```python
doces = int(input("Quantos doces estao disponiveis?"))
kids = int (input("Quantas crianças estão ai?"))
if doces < kids:
  print("Quantidade de doces insuficiente")
elif doces > kids:
  decu = doces - kids
  print(f"quantidade suficiente, sobrou {decu} doces")
else:
  print("Quantidade suficiente")
```

# Exercicio 10 — Reajuste salarial
* Crie uma variavel com o valor de um salario atual. Calcule um reajuste de 8% sobre esse valor
usando operadores aritmeticos e armazene o novo salario em outra variavel. Imprima o salario
antigo, o valor do aumento e o novo salario.

```python
salario_atual = float(input("Digite o salário atual: "))

reajuste_percentual = 0.08  # 8%

valor_aumento = salario_atual * reajuste_percentual
novo_salario = salario_atual + valor_aumento

print(f"Salário antigo: R$ {salario_atual:.2f}")
print(f"Valor do aumento: R$ {valor_aumento:.2f}")
print(f"Novo salário: R$ {novo_salario:.2f}")
```

# Exercicio 12 — Atribuicao composta
* Crie uma variavel chamada saldo com o valor 500. Em seguida, simule tres operacoes financeiras
usando operadores de atribuicao composta (+=, -=, *=): um deposito de 150, um saque de 80 e
uma taxa de manutencao que reduz o saldo em 5%. Imprima o saldo apos cada operacao.

```python
saldo = 500
print(f"Saldo inicial: R$ {saldo:.2f}")


saldo += 150
print(f"Saldo após depósito de R$ 150.00: R$ {saldo:.2f}")


saldo -= 80
print(f"Saldo após saque de R$ 80.00: R$ {saldo:.2f}")


saldo *= 0.95
print(f"Saldo após taxa de manutenção (5%): R$ {saldo:.2f}")
```

# Exercicio 13 — Conversao de tempo
* Crie uma variavel com uma quantidade de segundos (numero inteiro). Calcule quantos minutos e
segundos restantes essa quantidade representa, usando divisao inteira (//) e resto (%). Imprima o
resultado no formato: 185 segundos equivalem a 3 minutos e 5 segundos

```python
segundos_totais = int(input("Digite a quantidade de segundos: "))

minutos = segundos_totais // 60
segundos_restantes = segundos_totais % 60

print(f"{segundos_totais} segundos equivalem a {minutos} minutos e {segundos_restantes} segundos")
```

# Exercicio 15 — Ficha tecnica de um produto com calculo de imposto
* Crie variaveis para nome_produto, preco_base (float) e taxa_imposto (float, represente como 0.12
para 12%). Calcule o valor do imposto (preco_base multiplicado pela taxa) e o preco final
(preco_base mais o imposto). Use tambem um operador relacional para verificar se o preco final
ultrapassa 100 reais. Imprima o nome do produto, o valor do imposto, o preco final e o resultado
da verificacao.

```python
nome_produto = input("Digite o nome do produto: ")
preco_base = float(input("Digite o preço base do produto: "))
taxa_imposto = float(input("Digite a taxa de imposto (ex: 0.12 para 12%): "))

valor_imposto = preco_base * taxa_imposto
preco_final = preco_base + valor_imposto

preco_final_ultrapassa_100 = preco_final > 100

print(f"Nome do Produto: {nome_produto}")
print(f"Valor do Imposto: R$ {valor_imposto:.2f}")
print(f"Preço Final: R$ {preco_final:.2f}")
print(f"O preço final ultrapassa R$ 100? {preco_final_ultrapassa_100}")
```
